```python
s=np.ones(5)
print(s)
```

```python
s=np.ones((3,4))
print(s)
```

```python
s=np.zeros(8)
print(s)
```

```python
#random
s=np.random.random(5)
print(s)
```

```python
s=np.random.uniform(10,50,5)
print(s)
```

```python
s=np.random.rand(5)
print(s)
print(s.mean())
print(s.std())
```

```python
s=np.array([10,20,30,40,50,60])
result=np.random.choice(s)
print(result)
```

```python
s=np.array([10,20,30,40,50,60])
result=np.random.choice(s,2)
print(result)
print(type(result))
```

```python
# random.choice with probablity which favours one element in specified
s=[10,20,30]
result=np.random.choice(s,size=3,p=[0.1,0.6,0.3])
print(result)
```

```python
# to freez the results we can use seed
s=np.random.random(5)
np.random.seed(3)
print(s)
```

```python
#shuffle
s=np.array([1,2,3,4,5])
result=np.random.shuffle(s)
print(result)
print(s)
print(type(s))
```

```python
# permutation
s=np.array([1,2,3,4,5])
result=np.random.permutation(s)
print(s)
print(result)
```
